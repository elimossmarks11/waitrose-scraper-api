```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd

URL = "https://www.waitrose.com/ecom/shop/browse/groceries/pet"

```

```python
# define a dataframe to store data
df = pd.DataFrame()
```

```python
driver = webdriver.Chrome()

# Go to a different page
driver.get(URL)
driver.execute_script("document.body.style.zoom='0.5'")
print(f"Loaded Waitrose page. Page title: {driver.title}")
```

```python
while True:
    try: 
        element = WebDriverWait(driver, 3).until(
            EC.presence_of_element_located((By.CSS_SELECTOR, "button[data-testid='reject-all']"))
        )
        element.click()
        print("Cookie banner closed.")
        break
    except:
        pass 
```

```python
def load_all_products(driver):
    while True:
        try:
            # Wait until the 'Load More' button is clickable
            # https://selenium-python.readthedocs.io/waits.html

            element = WebDriverWait(driver, 5).until(
                EC.element_to_be_clickable((By.CSS_SELECTOR, "button[data-shortdescription='Load more products']"))
            )

            # load_more_button = driver.find_element(By.CSS_SELECTOR, "button[data-shortdescription='Load more products']")
            element.click()
            print("Clicked 'Load More' button.")
    
        except Exception as e:
            print("No more 'Load More' button found or an error occurred:", e)
            break 

load_all_products(driver)
```

```python
containers = driver.find_elements(By.CSS_SELECTOR, ".content___yamWw")

# print the number of containers found
print(f"Number of containers found: {len(containers)}")
```

```python
product_names = []

for container in containers:
    try:
        # 1. target the h2 using its stable data-testid
        name_element = container.find_element(By.CSS_SELECTOR, "h2[data-testid='product-pod-name']")
        
        # 2. extract the text
        product_name = name_element.text
        product_names.append(product_name)

        print(f"Extracted and stored in a dataframe: {product_name}")
        
    except Exception as e:
        print(f"Could not find name in this container")

df['Product Name'] = product_names
```
