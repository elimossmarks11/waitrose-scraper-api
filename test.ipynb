```python
import requests

barcodes_960 = ["5000116112111","5000116107568","5000116110810","5000116127795","5000116126149"]
barcodes_500 = ["5000116107544","5000116110995","5000116112104"]

for barcode in barcodes_500:
    url = f"https://world.openfoodfacts.org/api/v2/product/{barcode}?fields=product_name,nova_group"
    headers = {
        'User-Agent': 'MyCoolNutritionApp/1.0 (contact@example.com)'
    }

    response = requests.get(url, headers=headers)
    data = response.json()
    product = data.get("product", {})

    if data.get("status") == 1:
        product = data["product"]
        print(f"500 Product Name: {product.get('product_name')}")
        print(f"500 Nova Group: {product.get('nova_group')}")
    else:
        print("Product not found.")

for barcode in barcodes_960:
    url = f"https://world.openfoodfacts.org/api/v2/product/{barcode}?fields=product_name,nova_group"
    headers = {
        'User-Agent': 'MyCoolNutritionApp/1.0 (contact@example.com)'
    }

    response = requests.get(url, headers=headers)
    data = response.json()
    product = data.get("product", {})

    if data.get("status") == 1:
        product = data["product"]
        print(f"Product Name: {product.get('product_name')}")
        print(f"Nova Group: {product.get('nova_group')}")
    else:
        print("Product not found.")
```
